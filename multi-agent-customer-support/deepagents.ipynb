# Deep agents is an agent harness . 
It is the same core tool calling loop as other agent frameworks, but with built in capabilities that makes the agents reliable for the real tasks.


deepagents is a standalone library built on top of langchains core building blocks for agents. 

It uses the langgraph runtime for durable execution, streaming, human0in-loop and the other features.

Langchain is the framework that provides the core building blocks for your agents.



Langchain maintains several open source packages to help you build agents. Each serves a different purpose in the agent development stack.

We need to first understand the difference between the 

1. Agent frameworks
2. Agent Runtimes
3. Agent Harness 


# 1. Agent Frameworks (Langchain)

Agent frameworks provide the abstaction that makes it easier to someone to get started when building with LLM's

Langchain is an agent framework that provides abstractions like structured content blocks, the agent lopp and middleware.

While langchain is built on top of the LangGraph, you dont need to know langgraph to use langchain.

# 2. Agent Runtimes(langgraph)
 Agent runtimes it will provides the tooling for running agents in production, Supported tools are these

 1. *Durable execution* : Agents persistent through failures and can run for extended periods, resuming from where they left off.
 2. *Streaming*: Support for streaming workflows and responses.
 3. *HUman-in-the-loop *: It helps to incoroporate human oversight by inspecting and modifying agent state.
 4. *Persistance*:  Threat level and cross thread persistance for the state management.
 5. *Low level control*: Direct control over agent orchestration without high level abstractions.


Langgraph is a low level orchestration framework and runtime for building the agents, managing them and deploying long running stateful agents.

Agent frameworks are generally hiher level and run on agent runtimes. For examples langchain 1.0 is built on top of the langgraph.

Other examples of the agent runtimes are:
1. Temporal
2. Inngest

when to use langgraph 

use langgraph when:
1.  When we need fine grained, low level control over the agent orchestraion.
2. You nedd durabe execution for long running stateful agents.
3. Youre building comples workflows that combines deterministic and agentic loops.
4. You need a production readi infrastructure for the agent deployment.

# 3.  Agent Harnness( Deep agents SDK)
 Agent harnness are really opinion based, they provide powered frameworks with buil in tools and capabilities for building sophisticated long running agents. Supported tools may include:
 
 1. Planning capabilitie: Track multiple taks with a to-do-list.
 2. Task delegation : delegate works and to keep context clean with subagents.
 3. File system: Read and write access to files on different pluggable storage  backeends.
 4. Token managment: Conversation history summarization and for large tool result eviction.

 The Deep agents SDK is built on top of langgraph and it will add a planning capabilites, file systems for context management, and also the ability to spwan the subagents and more. Deep agents is designed for complex multistep tasks and required planning and decomposition.
