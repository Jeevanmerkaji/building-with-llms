Why Docker? 
Becasue It runs on my machine irrespective of what machine i am working and on what machine the application was developed.

Docker was born so that the application will not break on the production.

Docker applications into standardized isolated units called containers.

We can separate different parts of the code like agents application , database, main application as isolated containers so that it will run independently.


Vetical  Scaling -- Even u have the best computer there will be somelimit beyond which u can scale.
Horizontal scaling --- Docker helps us here in horizontal scaling say here we will be 100's of machines which will have thier independent memories and GPU's which will not shared with the other application. 


what is an image and what is a container?

The blueprint how the application to run. This blueprint can be used to make any number of application. 

For example you have made a blueprint fot a building and now you can use this blueprint to make any number of buildings, and the buildings here are nothing but the containers.


In python you can imagine the image as a class and the container as the object that means you can any number of the objects of the same class. 

Docker File --  It will contain all the recipe and the ingredient list that is required to make a docker image.

Docker Image -- And this is the blue print for making the product which will follow the ingredients and the recipe.

Docker Container --  It is the final product.
