# Use of LiteLLM — a mainstream, open-source LLM gateway
LiteLLM gives you one unified, OpenAI-compatible API across 100+ LLM providers (Anthropic, OpenAI, etc).
It can be used as a Python SDK (`litellm.completion(...)`) or run as a standalone proxy server.
These are its main features:
1. Chat completions
2. Streaming
3. Tool / function calling
4. Fallbacks and retries across providers
5. Caching, rate limiting, and cost tracking built in

With this, we don't need a separate framework per provider — one call shape works everywhere.

```python
# Portkey implementation
from dotenv import load_dotenv
import os

load_dotenv()
from portkey_ai import Portkey

portkey =  Portkey(
    api_key= os.environ["PORTKEY_API"]
)

response  = portkey.chat.completions.create(
    model = "@abc/claude-haiku-4-5",
    messages = [
      {"role": "system", "content": "You are a helpful assistant."},
      {"role": "user", "content": "What is Portkey"}
    ],
    max_tokens = 512
)

print(response.choices[0].message.content)
```

# Feature 2 -- Metadata and Observability

Tag every request with the user session and the feature info.

There are certain questions that we can answer from doing this and that is 
1. which user is generating the most cost 
2. Which feature uses the most tokens.
3. Which pipeline is slower is it the RAG pipeline or the support bot

```python
portkey = Portkey(
    api_key=os.environ["PORTKEY_API"]
)

response = portkey.with_options(
    metadata={
        "_user": "jeevan",
        "session_id": "this_session_abc",
        "feature": "Rag_application",
        "environment": "production",
    }
).chat.completions.create(
    model="@groq/qwen/qwen3.6-27b",
    messages=[
        {
            "role": "user",
            "content": "What is kubernetes"
        }
    ]
)

print(response.choices[0].message.content)
```

# Automatic Retries 

I need to pass the config dict to the Portkey(). On a transient error(rate limit, server error), Portkey retries automaically with exponential backoff. Basically what happens is that the app newer sees this kind of failures


```python
portkey = Portkey(api_key= os.environ["PORTKEY_API"],
                  config ={
                      "retry" : {
                          "attempts" : 3,
                          "on_status_codes" : [429, 500, 502, 503, 504]
                      }
                  })
```

# Feature 4 - Request Timeouts

We need to set a hardlimit in order to limit the time the portkey will wait before returning the HTTP 408 response. We set this time limit in milliseconds. If the LLM stalls then the portkey kill the request. Without this what will happen is that the slow Groq response will actually block the FASTApi worker indifinitely.






```python
portkey = Portkey(api_key= os.environ["PORTKEY_API"], 
                  config = {
                      "request_timeout" : 10000 # 10 sec
                  })



## This can be combined also with the retries 
config = {
    "request_timeout": 10000,
    "retry": {"attempts": 2, "on_status_codes": [408, 429, 503]}
}
```

# Feature -5 Fallbacks

If the primary model fails, Portkey will automatically switches to the next target in the list. Users never see the failure



```python
portkey = Portkey (api_key= os.environ["PORTKEY_API"] , 
                   config={
                       "strategy" : {"mode" : "fallback"},
                        "targets": [
                            {"override_params": {"model": "@groq/llama-3.3-70b-versatile"}},  # primary
                            {"override_params": {"model": "@groq/llama-3.1-8b-instant"}}       # fallback
                        ]
                       
                   })


"""
    Narrowing the trigger (avoids fallback on bad requests):
    "strategy": {"mode": "fallback", "on_status_codes": [429, 503]}

"""
```
