# LLM gateway it sits between your application and the API providers and successfu;;y handle these main things 
1. Routing
2. Fallbacks
3. Caching
4. Rate Limiting
5. Gaurdrails
6. Cost Tracking

These are few first thing  which are very important to know. 
1.  Unified API - Here we are going to define the pone funnction and we will be easily switch between the apis.
2. Automatic Fallback  -- usage of the backup models
3. Smart Routing -- based on the different requests coming to the gateways ..it will direct it according to the required api 
4. Load Balancing  -- It will switch between the models based on the load that the model can handle
5. Caching -- there may be 100 same questions that are being asked by the user so the LLM now knows that what kind of questions are being asked repeatedly and it will be able to make decisions quickly by caching
6. Observability  -- every call that will be happening is being logged we can actually see the every $ where it is spent
7. Gaurdrails -- Depending on the info which we give such as credit card numbers or some sensitive information that can be stopped from reaching the LLM providers and can be filtered by setting the rules and that is by using the Gaurdrails 
8. Evals -- We can also set some evaluation criteria being set in order to evaluate the entire process.
